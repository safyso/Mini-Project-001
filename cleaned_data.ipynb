```python
import pandas as pd
import numpy as np
import pyarrow

class DataCleaner:
    def __init__(self, file_path):
        self.file_path = file_path
        self.data = None

    def load_data(self):
        # Setup: Load the raw dataset
        try:
            self.data = pd.read_excel(self.file_path)
            print(f"Data loaded successfully. Shape: {self.data.shape}")
        except Exception as e:
            print(f"Error loading data: {e}")

    def drop_unnecessary_columns(self):
        # Getting Oriented: Drop columns that are not needed for analysis
        try:
            cols_to_drop = ["Row ID"]
            existing_cols = [col for col in cols_to_drop if col in self.data.columns]
            if existing_cols:
                self.data = self.data.drop(columns=existing_cols)
            print("Unnecessary columns dropped.")
        except Exception as e:
            print(f"Error dropping columns: {e}")

    def clean_text_columns(self):
        # Inconsistent Values: Strip extra spaces and convert to title case
        try:
            string_cols = self.data.select_dtypes("object").columns.tolist()
            for col in string_cols:
                self.data[col] = self.data[col].astype(str).str.strip().str.title()
            print("Text columns cleaned and standardized.")
        except Exception as e:
            print(f"Error cleaning text columns: {e}")

    def convert_data_types(self):
        # Data Types & Conversions: Convert dates and optimize memory with categories
        try:
            date_cols = ["Order Date", "Ship Date"]
            for col in date_cols:
                if col in self.data.columns:
                    self.data[col] = pd.to_datetime(self.data[col])
            
            string_cols = self.data.select_dtypes("object").columns.tolist()
            for col in string_cols:
                if self.data[col].nunique() <= 50:
                    self.data[col] = self.data[col].astype("category")
            print("Data types optimized.")
        except Exception as e:
            print(f"Error converting data types: {e}")65225

    def handle_missing_data(self):
        # Missing Data: Fill missing values in Postal Code
        try:
            if "Postal Code" in self.data.columns:
                self.data["Postal Code"] = self.data["Postal Code"].astype(str)
                self.data["Postal Code"] = self.data["Postal Code"].replace(['Nan', 'nan', 'NaN'], "05401")
            print("Missing data handled.")
        except Exception as e:
            print(f"Error handling missing data: {e}")

    def remove_duplicates(self):
        # Duplicate Detection: Remove fully duplicated rows
        try:
            initial_rows = self.data.shape[0]
            self.data = self.data.drop_duplicates()
            final_rows = self.data.shape[0]
            print(f"Removed {initial_rows - final_rows} duplicated rows.")
        except Exception as e:
            print(f"Error removing duplicates: {e}")

    def handle_outliers(self, columns=['Sales', 'Profit']):
        # Outliers: Flag extreme values per Segment using the IQR method without dropping them
        try:
            self.data['Outlier_Flag'] = False
            
            for col in columns:
                if col in self.data.columns:
                    Q1 = self.data.groupby('Segment')[col].transform(lambda x: x.quantile(0.25))
                    Q3 = self.data.groupby('Segment')[col].transform(lambda x: x.quantile(0.75))
                    IQR = Q3 - Q1
                    
                    lower_bound = Q1 - 3 * IQR
                    upper_bound = Q3 + 3 * IQR
                    
                    mask = (self.data[col] < lower_bound) | (self.data[col] > upper_bound)
                    self.data.loc[mask, 'Outlier_Flag'] = True
                    
                    outliers_count = mask.sum()
                    print(f"Flagged {outliers_count} potential outliers in {col} (Calculated per Segment).")
        except Exception as e:
            print(f"Error handling outliers: {e}")

    def export_cleaned_data(self, output_name="Sample - Superstore 2019_clean.parquet"):
        # Exporting: Save the cleaned DataFrame to a Parquet file
        try:
            if self.data is not None:
                self.data.to_parquet(output_name, engine="pyarrow", index=False)
                print(f"Data successfully exported to: {output_name}")
        except Exception as e:
            print(f"Error exporting data: {e}")

    def run_cleaning_pipeline(self):
        # Master Pipeline: Executing all steps sequentially
        self.load_data()
        
        if self.data is not None:
            self.drop_unnecessary_columns()
            self.clean_text_columns()
            self.convert_data_types()
            self.handle_missing_data()
            self.remove_duplicates()
            self.handle_outliers()
            self.export_cleaned_data()
            
            print("Cleaning pipeline executed successfully.")
            return self.data
        else:
            print("Pipeline aborted: Data not loaded.")
            return None


# --- Execution ---

# 1. Create an object from the DataCleaner class
cleaner = DataCleaner("Sample - Superstore 2019.xls")

# 2. Run the full cleaning pipeline and store the result in df_clean
df_clean = cleaner.run_cleaning_pipeline()

# 3. Quick check on the final data
if df_clean is not None:
    print("\n--- Final Data Info ---")
    df_clean.info()
```
